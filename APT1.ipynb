```python

from selenium import webdriver
import time
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options


options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)

 


driver = webdriver.Chrome(options=options)

driver.get("https://www.naukrigulf.com")

time.sleep(3)


driver.find_element(By.CLASS_NAME, 'qsb-proxy').click()

time.sleep(3)


driver.find_element(By.XPATH, '//*[@id="qsbKey"]').send_keys("Data Engineer")

time.sleep(3)


driver.find_element(By.CLASS_NAME, "blue-retro").click()

time.sleep(5)
```

```python
import pandas as pd
```

```python
all_jobs = []

# nget all job cards using a more general XPath that captures different structures
job_cards = driver.find_elements(By.XPATH, "//div[contains(@class, 'job') or contains(@class, 'tuple') or contains(@class, 'card')]")

for card in job_cards:
    # 1. Job Title
    try:
        job_title = card.find_element(By.TAG_NAME, 'a').text
    except:
        job_title = "N/A"
        
    # 2. Company Name
    try:
        company_name = card.find_element(By.XPATH, ".//div[contains(@class, 'comp') or contains(@class, 'org')]").text
    except:
        company_name = "N/A"
        
    # 3. Location
    try:
        location = card.find_element(By.XPATH, ".//span[contains(@class, 'loc')] | .//li[contains(@class, 'loc')]").text
    except:
        location = "N/A"
        
    # 4. Experience
    try:
        experience = card.find_element(By.XPATH, ".//span[contains(@class, 'exp')] | .//li[contains(@class, 'exp')]").text
    except:
        experience = "N/A"
        
    # 5. Full Job Description
    try:
        description = card.find_element(By.XPATH, ".//div[contains(@class, 'desc')] | .//p").text
    except:
        description = "N/A"

    # nadd the job details to the list
    all_jobs.append({
        'Job Title': job_title,
        'Company Name': company_name,
        'Job Location': location,
        'Required Experience': experience,
        'Full Job Description': description
    })

print("done ، number of jobs :", len(all_jobs))
```

```python
import time
from selenium.webdriver.common.by import By

for page in range(2, 4):
    try:
        # nscroll to the bottom of the page to load more job cards
        driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
        time.sleep(2)

        # ngo to next page
        driver.get(f"https://www.naukrigulf.com/data-engineer-jobs-{page}")
        time.sleep(5)

        # nget all job cards on the new page
        job_cards = driver.find_elements(By.XPATH, "//div[contains(@class, 'job') or contains(@class, 'tuple') or contains(@class, 'card')]")

        for card in job_cards:
            # 1. Job Title
            try:
                job_title = card.find_element(By.TAG_NAME, 'a').text
            except:
                job_title = "N/A"
                
            # 2. Company Name
            try:
                company_name = card.find_element(By.XPATH, ".//div[contains(@class, 'comp') or contains(@class, 'org')]").text
            except:
                company_name = "N/A"
                
            # 3. Location
            try:
                location = card.find_element(By.XPATH, ".//span[contains(@class, 'loc')] | .//li[contains(@class, 'loc')]").text
            except:
                location = "N/A"
                
            # 4. Experience
            try:
                experience = card.find_element(By.XPATH, ".//span[contains(@class, 'exp')] | .//li[contains(@class, 'exp')]").text
            except:
                experience = "N/A"
                
            # 5. Full Job Description
            try:
                description = card.find_element(By.XPATH, ".//div[contains(@class, 'desc')] | .//p").text
            except:
                description = "N/A"

            # nadd the job details to the list
            all_jobs.append({
                'Job Title': job_title,
                'Company Name': company_name,
                'Job Location': location,
                'Required Experience': experience,
                'Full Job Description': description
            })

    except Exception as e:
        print(f"Error on page {page}: {e}")
        break

print(f"all jobs: {len(all_jobs)}")
```

```python
df = pd.DataFrame(all_jobs)
df.to_csv("naukrigulf_jobs.csv", index=False)
```
